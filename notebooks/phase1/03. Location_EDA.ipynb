```python
1. Features por location × revenue center
toma analysis_master (una fila por check) y la resume en una fila por restaurante y área de venta. Calcula métricas de comportamiento como ticket promedio, items por check, % de descuento, % de propina, tasa de voids, checks sin pago (comps), checks vacíos y duración promedio. Usa try_cast para tolerar valores "null" en texto y abs() en la propina porque viene negativa. Descarta combinaciones con menos de MIN_CHECKS checks para no comparar contra ruido.
```

```python
from pyspark.sql import functions as F

am = spark.table("workspace.default.applebees_analysis_master")

LOC, LOCNAME, RC = "LocationId", "LocationName", "RevenueCenterName"
MIN_CHECKS = 115   # fechas de operación

def num(c):   # texto "null" -> 0.0
    return F.coalesce(F.expr(f"try_cast({c} AS DOUBLE)"), F.lit(0.0))

def flag(c):  # funciona si la columna es boolean o texto "TRUE"/"FALSE"
    return (F.upper(F.col(c).cast("string")) == "TRUE").cast("int")

# Duración en minutos (soporta serial tipo Excel o timestamp)
if dict(am.dtypes)["CloseTime"] in ("timestamp", "timestamp_ntz"):
    dur = (F.col("CloseTime").cast("long") - F.col("OpenTime").cast("long")) / 60
else:
    dur = (num("CloseTime") - num("OpenTime")) * 1440

base = (am
    .withColumn("sales",   num("total_sold_price"))
    .withColumn("tip",     F.abs(num("gratuity_raw")))
    .withColumn("disc",    num("discount_amount"))
    .withColumn("voidamt", num("void_amount"))
    .withColumn("paid",    (num("payment_amount") > 0).cast("int"))
    .withColumn("dur_min", dur)
)

features = (base.groupBy(LOC, LOCNAME, RC).agg(
        F.countDistinct("CheckId").alias("num_checks"),
        F.sum("sales").alias("sum_sales"),
        F.sum("disc").alias("sum_disc"),
        F.sum("voidamt").alias("sum_void"),
        F.sum(F.col("tip") * F.col("paid")).alias("sum_tip_paid"),
        F.sum(F.col("sales") * F.col("paid")).alias("sum_sales_paid"),
        F.avg(F.when(F.col("sales") > 0, F.col("sales"))).alias("avg_check"),
        F.avg(F.when(F.col("sales") > 0, num("total_item_quantity"))).alias("items_per_check"),
        F.avg((num("void_count") > 0).cast("int")).alias("void_rate"),
        F.avg(flag("flag_no_payments")).alias("comp_rate"),
        F.avg(flag("flag_empty_check")).alias("empty_rate"),
        F.avg(flag("flag_zero_duration")).alias("zero_dur_rate"),
        F.avg(F.when(F.col("dur_min").between(0, 360), F.col("dur_min"))).alias("avg_duration_min"),
    )
    .withColumn("discount_pct", F.expr("try_divide(sum_disc, sum_sales + sum_disc)"))
    .withColumn("void_pct",     F.expr("try_divide(sum_void, sum_sales)"))
    .withColumn("tip_pct",      F.expr("try_divide(sum_tip_paid, sum_sales_paid)"))
    .filter(F.col("num_checks") >= MIN_CHECKS)
)

metrics = ["avg_check", "items_per_check", "discount_pct", "tip_pct",
           "void_rate", "void_pct", "comp_rate", "empty_rate",
           "zero_dur_rate", "avg_duration_min"]

display(features.orderBy(RC, LOC).limit(20))
```

```python
2. Z-score robusto contra pares
Pasa la tabla (ya pequeña) a pandas y dentro de cada revenue center, calcula cuántas desviaciones se aleja cada restaurante de la mediana en cada métrica. Usa mediana y MAD en lugar de media y desviación estándar, para que los propios atípicos no distorsionen la comparación. Si el MAD es 0 (común en tasas que casi siempre valen 0), usa la desviación media absoluta como respaldo. Un |z| > 3.5 se marca como atípico. Guarda top_driver (la métrica que más lo desvía) y why (las 3 principales con su signo).
```

```python
import numpy as np
import pandas as pd

pdf = features.toPandas()

def robust_z(s):
    s = s.astype(float)
    med = s.median()
    dev = (s - med).abs()
    mad = dev.median()
    if mad > 0:
        return 0.6745 * (s - med) / mad
    mean_ad = dev.mean()   # respaldo si MAD = 0
    return (s - med) / (1.253314 * mean_ad) if mean_ad > 0 else s * 0.0

z_cols = []
for m in metrics:
    pdf[f"{m}_rz"] = pdf.groupby(RC)[m].transform(robust_z)
    z_cols.append(f"{m}_rz")
pdf[z_cols] = pdf[z_cols].fillna(0)

pdf["max_abs_z"]    = pdf[z_cols].abs().max(axis=1)
pdf["top_driver"]   = pdf[z_cols].abs().idxmax(axis=1).str.replace("_rz", "", regex=False)
pdf["is_outlier_z"] = pdf["max_abs_z"] > 3.5

def top_drivers(row, k=3):
    z = row[z_cols].astype(float)
    top = z.abs().sort_values(ascending=False).head(k).index
    return ", ".join(f"{c.replace('_rz', '')} ({z[c]:+.1f})" for c in top)

pdf["why"] = pdf.apply(top_drivers, axis=1)

print(pdf.groupby(RC).size())   # cuántos pares hay por revenue center
```

```python
3. Isolation Forest (detección multivariada)
El z-score revisa cada métrica por separado, mientras que Isolation Forest mira todas a la vez y detecta combinaciones raras que ninguna métrica sola mostraría. Se entrena por separado en cada revenue center y solo si hay al menos 20 pares. Al final, un restaurante es atípico si lo marcó el z-score o el modelo.
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import RobustScaler

parts = []
for rc, g in pdf.groupby(RC):
    g = g.copy()
    if len(g) < 20:                       # muy pocos pares para el modelo
        g["anomaly_score"], g["is_outlier_if"] = np.nan, False
    else:
        X = RobustScaler().fit_transform(g[metrics].fillna(0))
        iso = IsolationForest(contamination=0.05, random_state=42).fit(X)
        g["anomaly_score"] = iso.decision_function(X)   # más bajo = más raro
        g["is_outlier_if"] = iso.predict(X) == -1
    parts.append(g)

final = pd.concat(parts, ignore_index=True)
final["is_outlier"] = final["is_outlier_z"] | final["is_outlier_if"]
```

```python
5. Lista de restaurantes atípicos
Muestra solo los restaurantes marcados como atípicos, ordenados del más extremo al menos extremo, con la columna why. Esa columna resume en qué métricas se desvían y en qué dirección (+ arriba de sus pares, - abajo). También imprime cuántos hay por revenue center y qué métrica los causa más seguido.
```

```python
final["label"] = final[LOC].astype(str) + " · " + final[LOCNAME].astype(str)
out = final[final["is_outlier"]].sort_values("max_abs_z", ascending=False).copy()

print(f"{len(out)} atípicos de {len(final)} combinaciones location × revenue center "
      f"({len(out) / max(len(final), 1):.1%})")
print("\nAtípicos por revenue center:")
print(out.groupby(RC).size().to_string())
print("\nMétrica que más los causa:")
print(out["top_driver"].value_counts().to_string())

display(out[["label", RC, "num_checks", "max_abs_z",
             "is_outlier_z", "is_outlier_if", "anomaly_score", "why"]])
```

```python
sel = out[out["max_abs_z"] > 3.5].head(15)
rows = []
for _, r in sel.iterrows():
    peers = final[final[RC] == r[RC]]
    for zc in r[z_cols].astype(float).abs().sort_values(ascending=False).head(3).index:
        m = zc.replace("_rz", "")
        rows.append((r["label"], r[RC], m, r[m], peers[m].median()))
display(pd.DataFrame(rows, columns=["restaurante", "revenue_center", "metrica", "valor", "mediana_pares"]))
```

```python
6. Gráficas: quiénes son y por qué
Genera cuatro gráficas.
- Highlights: cuántos atípicos hay por revenue center y qué métrica los explica más.
- Ranking: los atípicos ordenados por su desviación máxima. La línea roja es el umbral (3.5), y los que quedan debajo los detectó el Isolation Forest, no el z-score.
- Mapa de calor: el z-score de cada métrica por restaurante (rojo = muy arriba de sus pares, azul = muy abajo). Muestra de un vistazo qué métrica rompe el patrón.
- Detalle: para los 6 más extremos, un histograma de sus pares en sus 3 métricas más desviadas, con una línea roja donde está el restaurante y una gris con la mediana.
```

```python
import matplotlib.pyplot as plt

TOP_N = 15
rc_colors = dict(zip(sorted(final[RC].unique()), plt.cm.tab10.colors))

# ---- 6a. Highlights ----
if len(out):
    fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
    cnt = out.groupby(RC).size()
    axes[0].bar(cnt.index, cnt.values, color=[rc_colors[r] for r in cnt.index])
    axes[0].set_title("Atípicos por revenue center")
    axes[0].set_ylabel("# restaurantes")
    axes[0].tick_params(axis="x", rotation=30)
    drv = out["top_driver"].value_counts().sort_values()
    axes[1].barh(drv.index, drv.values, color="#4C78A8")
    axes[1].set_title("Métrica que más explica los atípicos")
    plt.tight_layout(); plt.show()

# ---- 6b. Ranking ----
top = out.head(TOP_N).iloc[::-1]
if len(top):
    fig, ax = plt.subplots(figsize=(10, max(3, 0.45 * len(top) + 1.5)))
    ylabels = top["label"] + "  |  " + top[RC]
    ax.barh(ylabels, top["max_abs_z"], color=[rc_colors[r] for r in top[RC]])
    ax.axvline(3.5, ls="--", c="red", label="umbral z = 3.5")
    ax.set_xlabel("Desviación robusta máxima (|z|)")
    ax.set_title(f"Top {len(top)} restaurantes atípicos")
    handles = [plt.Rectangle((0, 0), 1, 1, color=c) for c in rc_colors.values()]
    ax.legend(handles + [ax.lines[0]], list(rc_colors) + ["umbral z = 3.5"], loc="lower right")
    plt.tight_layout(); plt.show()

# ---- 6c. Mapa de calor del "por qué" ----
hm = out.head(TOP_N)
if len(hm):
    vals = hm[z_cols].values.astype(float)
    fig, ax = plt.subplots(figsize=(12, 0.5 * len(hm) + 2.5))
    im = ax.imshow(np.clip(vals, -6, 6), cmap="RdBu_r", vmin=-6, vmax=6, aspect="auto")
    ax.set_xticks(range(len(metrics)))
    ax.set_xticklabels(metrics, rotation=45, ha="right")
    ax.set_yticks(range(len(hm)))
    ax.set_yticklabels(hm["label"] + " | " + hm[RC], fontsize=8)
    for i in range(vals.shape[0]):
        for j in range(vals.shape[1]):
            if abs(vals[i, j]) >= 3.5:
                ax.text(j, i, f"{vals[i, j]:+.1f}", ha="center", va="center",
                        fontsize=8, fontweight="bold")
    plt.colorbar(im, label="z robusto (rojo = arriba de pares, azul = abajo)")
    ax.set_title("¿En qué métricas se desvía cada atípico?")
    plt.tight_layout(); plt.show()

# ---- 6d. Detalle: atípico vs. sus pares ----
N = min(6, len(out))
if N:
    fig, axes = plt.subplots(N, 3, figsize=(14, 2.7 * N), squeeze=False)
    for i, (_, r) in enumerate(out.head(N).iterrows()):
        peers = final[final[RC] == r[RC]]
        drivers = r[z_cols].astype(float).abs().sort_values(ascending=False).head(3).index
        for j, zc in enumerate(drivers):
            m = zc.replace("_rz", "")
            ax = axes[i][j]
            ax.hist(peers[m].dropna(), bins=20, color="lightgray", edgecolor="white")
            ax.axvline(peers[m].median(), color="gray", ls="--",
                       label="mediana pares" if (i == 0 and j == 0) else None)
            ax.axvline(r[m], color="red", lw=2,
                       label="este restaurante" if (i == 0 and j == 0) else None)
            ax.set_title(f"{r['label']} | {r[RC]}\n{m}  (z = {r[zc]:+.1f})", fontsize=9)
    fig.legend(loc="upper right")
    plt.tight_layout(); plt.show()
```
